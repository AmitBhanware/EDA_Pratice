```python
try:
  # exception can occur
  a = int(input('Enter the number 1 '))
  b = int(input('Enter the number 2 '))
  c = a/b
  d = a*b
  e = a+b
  print(c)
  print(d)
  print(e)

except NameError:
  print('User has not defined the variable')

except ZeroDivisionError:
  print('give number greater than 0')
except TypeError:
  print('Try to make data type similar')

except Exception as ex :
  print(ex)

else:
  print(c)
  print(d)
  print(e)

finally:
  print('execution is done')
```

```python
class Error(Exception):
  pass

class dobException(Error):
  pass
```
