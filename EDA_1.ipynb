```python
df_country = pd.read_excel('Country-Code.xlsx')
```

```python
df_country.head()
```

```python
final_df = pd.merge(df, df_country, on = 'Country Code', how = 'left')
```

```python
final_df.head()
```

```python
country_names = final_df.Country.value_counts().index
```

```python
country_values = final_df.Country.value_counts().values
```

```python
# pie chart - Top 3 countries

plt.pie(country_values[:3], labels = country_names[:3], autopct = '%1.2f%%')
```

Obsevation :- Zomato maximum transaction are from India then USA and UK

```python
final_df.columns
```

```python
ratings =  final_df.groupby(['Aggregate rating', 'Rating color', 'Rating text']).size().reset_index().rename(columns = {0 : 'Rating count'})
```

Observation

1. when rating is between 4.5 to 4.9  ----> Excellent
2. when rating is between 4.0 to 3.4  ---> vary Good
3. when rating is between 3.5 to 3.9  ---> Good
4. when rating is between 3 to 3.4  ---> average
5. when rating is between 2.5 to 2.9  ---> average
6. others are poor

```python
ratings.head()
```

```python
import matplotlib
matplotlib.rcParams['figure.figsize'] = (12,6)
sns.barplot(x = 'Aggregate rating', y = 'Rating count', data =ratings, hue = 'Rating color', palette = ['blue','red','orange','yellow','green','purple'])
```

```python
sns.countplot(x = 'Rating color',data = ratings, palette = ['blue','red','orange','yellow','green','purple'] )
```
