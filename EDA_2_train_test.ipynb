```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt
```

```python
%matplotlib inline
```

```python
#importing tarain dataset

df_train = pd.read_csv('train.csv')
df_train.head()
```

```python
#import test datset

df_test = pd.read_csv('test.csv')
df_test.head()
```

```python
import pandas as pd

# append both data set
df = pd.concat([df_train, df_test], axis = 0)
df.head()
```

```python
df.info()
```

```python
df.describe()
```

```python
df.drop('User_ID', axis = 1 , inplace = True)
```

```python
df.head()
```

```python
df['Gender'] = df['Gender'].map({'F': 0, 'M': 1})
df.head()
```

```python
df['Gender'].unique()
```

```python
df['Age'].unique()
```

```python
df['Age'] = df['Age'].map({'0-17': 1, '18-25': 2, '26-35': 3, '36-45': 4,'46-50':5, '51-55':6, '55+':7})
```

```python
df['Age'].unique()
```

```python
from sklearn import preprocessing

label_encoder = preprocessing.LabelEncoder()

df['Age'] = label_encoder.fit_transform(df['Age'])

df['Age'].unique()
```

```python
df_city = pd.get_dummies(df['City_Category'], drop_first = True)
```

```python
df_city.head()
```

```python
df = pd.concat([df, df_city], axis = 1)
df.head()
```

```python
df.drop('City_Category', axis = 1, inplace = True)
```

```python
# Missing value
df.isnull().sum()
```

```python
# Fill mising value
df['Product_Category_1'].unique()
```

```python
df['Product_Category_2'].value_counts()
```

```python
# best way to replace missing value mode
# Product_Category_2
mode_product_category_2 = df['Product_Category_2'].mode()[0]
df['Product_Category_2'] = df['Product_Category_2'].fillna(mode_product_category_2).astype(int)
```

```python
df['Product_Category_2'].head()
```

```python
df['Product_Category_2'].isnull().sum()
```

```python
mode_product_category_3 = df['Product_Category_3'].mode()[0]
df['Product_Category_3'] = df['Product_Category_3'].fillna(mode_product_category_3).astype(int)
```

```python
df.head()
```

```python
df['Stay_In_Current_City_Years'] = df['Stay_In_Current_City_Years'].str.replace('+', '')
```

```python
df.columns
```

```python
df['B'] = df['B'].astype(int)
df['C'] =  df['C'].astype(int)
```
